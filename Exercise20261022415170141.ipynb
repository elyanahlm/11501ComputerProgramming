### Part 1：完成函數

請閱讀以下函數，理解各部分的功能。

```python
def analyze_scores(*scores, **options):
    """
    分析學生分數。

    Parameters:
        *scores: 接收不定數量的分數
        **options: 接收選擇性設定，例如 multiplier 和 bonus

    Returns:
        一個字典，包含調整後的分數、平方、總分及平均分數。
    """
    multiplier = options.get("multiplier", 1)
    bonus = options.get("bonus", 0)

    adjusted = [
        score * multiplier + bonus
        for score in scores
    ]

    squared = [
        score ** 2
        for score in adjusted
    ]

    total = sum(adjusted)
    average = total / len(adjusted) if adjusted else 0

    return {
        "adjusted_scores": adjusted,
        "squared_scores": squared,
        "total": total,
        "average": average
    }
```

## **Part 3**

1. 在函數內部，`scores` 的資料型態是什麼？其值為何？
scores 的資料型態是 tuple（元組）。當呼叫 analyze_scores(60, 70, 80) 時，*scores 會將所有位置參數收集成 tuple，因此其值為 (60, 70, 80)。

2. 在函數內部，`options` 的資料型態是什麼？其值為何？
options 的資料型態是 dict（字典）。當呼叫 analyze_scores(60, 70, 80, multiplier=1.1, bonus=5) 時，**options 會將所有關鍵字參數收集成 dictionary，因此其值為 {"multiplier": 1.1, "bonus": 5}。

3. `score * multiplier` 與 `score ** 2` 有什麼不同？請說明兩個運算子的意義。
「*」代表乘法運算，例如 60 * 1.1 = 66；「**」代表次方運算，例如 60 ** 2 = 3600。因此 * 用於乘法，** 用於計算次方。

4. 為什麼函數使用 `return`，而不是只使用 `print()`？
return 可以將函數的結果回傳給呼叫端，方便後續儲存及使用；而 print() 只能將結果顯示在螢幕上，無法提供其他程式碼進一步使用，因此 return 較具有彈性。

5. 如果呼叫 `analyze_scores(60, 70, 80)`，但沒有提供 `settings`，程式會如何執行？請說明原因。
程式仍可正常執行，因為 options.get() 提供了預設值。當沒有提供 multiplier 與 bonus 時，multiplier 會使用 1，bonus 會使用 0，因此程式會直接使用原始分數進行計算。

6. 請修改函數，讓回傳的字典額外包含調整後的最高分數，鍵值名稱為 `"max_score"`。
--> 可使用 max() 函數取得調整後分數中的最高分數，並在回傳的字典中加入 "max_score" 鍵值。

````python
def analyze_scores(*scores, **options):

    multiplier = options.get("multiplier", 1)
    bonus = options.get("bonus", 0)

    adjusted = [
        score * multiplier + bonus
        for score in scores
    ]

    squared = [
        score ** 2
        for score in adjusted
    ]

    total = sum(adjusted)
    average = total / len(adjusted) if adjusted else 0

    return {
        "adjusted_scores": adjusted,
        "squared_scores": squared,
        "total": total,
        "average": average,
        "max_score": max(adjusted)
    }
````

7. 如果沒有傳入任何分數，程式會回傳什麼結果？請解釋程式如何處理這種情況。
如果沒有傳入任何分數，scores 會是一個空的 tuple，因此 adjusted 和 squared 會是空串列，total 為 0。程式利用 if adjusted else 0 判斷是否有資料，避免除以零錯誤，因此 average 會回傳 0。


## **Part 4**

1. 原本的 `*scores` 與 `scores_list` 有什麼不同？
*scores 會將多個位置參數收集成 tuple，例如 analyze_scores(60, 70, 80)；而 scores_list 是一個關鍵字參數，將所有分數放在串列中傳入，例如 scores_list=[60, 70, 80]。兩者都能表示多筆分數，但傳遞方式不同。

2. 為什麼原本的函數無法直接使用上述呼叫方式處理 `scores_list`？
原本的函數只能透過 *scores 接收分數，因此 analyze_scores(60, 70, 80) 可以正常運作。當使用 scores_list=[60, 70, 80] 時，資料會被存入 options，而不是 scores。由於函數沒有讀取 options 中的 scores_list，因此無法直接處理該呼叫方式。

3. 請修改函數，使上述呼叫方式也能正確計算結果。
當沒有透過 *scores 傳入分數時，程式會使用 options.get("scores_list", []) 取得 scores_list 作為分數來源，因此函數同時支援 *scores 與 scores_list 兩種輸入方式。

````python
def analyze_scores(*scores, **options):

    multiplier = options.get("multiplier", 1)
    bonus = options.get("bonus", 0)

    if not scores:
        scores = options.get("scores_list", [])

    adjusted = [
        score * multiplier + bonus
        for score in scores
    ]

    squared = [
        score ** 2
        for score in adjusted
    ]

    total = sum(adjusted)
    average = total / len(adjusted) if adjusted else 0

    return {
        "adjusted_scores": adjusted,
        "squared_scores": squared,
        "total": total,
        "average": average
    }
````

*的用途

在 Python 中，* 有多種用途。

在運算式中，* 表示乘法。
在函數定義中，例如 *scores，可收集不定數量的位置參數。
在函數呼叫時，例如 *scores，可將串列或其他可迭代物件解包成多個位置參數。


**的用途

在 Python 中，** 也有多種用途。

在運算式中，** 表示次方運算。
在函數定義中，例如 **options，可收集不定數量的關鍵字參數。
在函數呼叫時，例如 **settings，可將字典解包成關鍵字參數。


return 的用途

return 用來將函數的計算結果回傳給呼叫端。

與 print() 不同，return 回傳的資料可以儲存於變數中，並在後續程式中重複使用或進一步處理，因此具有較高的彈性。


## 範例一：* 作為乘法運算

````python
score = 60
result = score * 2

print(result)
````

## 範例二：*args 收集不定數量的位置參數

````python
def show_scores(*scores):
    print(scores)

show_scores(60, 70, 80)
````

## 範例三：* 解包（Unpacking）串列

````python
def show_scores(a, b, c):
    print(a, b, c)

scores = [60, 70, 80]

show_scores(*scores)
````

## 範例四：** 作為次方運算

````python
score = 60

print(score ** 2)
````

## 範例五：**kwargs 收集不定數量的關鍵字參數

````python
def show_info(**info):
    print(info)

show_info(name="Anna", score=100)
````

## 範例六：** 解包（Unpacking）字典

````python
def student(name, score):
    print(name, score)

data = {
    "name": "Anna",
    "score": 100
}

student(**data)
````

## 範例七：return 回傳函數結果

````python
def add(a, b):
    return a + b

result = add(3, 5)

print(result)
````
