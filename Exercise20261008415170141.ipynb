使用者輸入：請輸入正整數$n$，其中$1\le n\le 9$，輸出$nn$乘法表。格式如Tranclass的圖片Exercise20261008.png，以$n=5$為例。

1. 只能使用for。
3. 字串格式化只能使用 f-string 對齊。

```python
n = int(input("請輸入 n："))

for row in range(1, n + 1):
    for col in range(1, n + 1):
        print(f"{row:03b}×{col:03b}={row*col:06b}", end ="\t")
    print()
```

使用者輸入：請輸入正整數$n$，其中$1\le n\le 9$，輸出二進位$nn$乘法表。格式如如Tranclass的圖片Exercise20261008.png，以$n=5$為例。                                                                                                         
                                                                                                            
1. 只能使用while。                                                                                            
3. 字串格式化只能使用 f-string 對齊。                                                                       

```python
n = int(input("請輸入 n："))

row = 1

while row <= n:
    col = 1

    while col <= n:
        print(f"{row:03b}×{col:03b}={row*col:06b}", end="  ")
        col += 1

    print()
    row += 1
```
