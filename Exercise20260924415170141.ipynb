## 第二部分：數學概念與 Python 型別

數學上有下列包含關係：

$$
\mathbb N
\subseteq
\mathbb Z
\subseteq
\mathbb Q

$$

請回答：

1. Python 是否為上述每一個數系都提供獨立型別？不是。 Python 並沒有為每個數學數系都提供獨立型別。其中自然數與整數共用 int，而有理數甚至沒有內建型別。
2. 為什麼自然數與整數都使用 `int` 表示？因為自然數只是整數的子集合。因此 Python 不需要另外設計一個「自然數型別」，直接用 int 表示所有整數即可。
3. Python 為什麼提供 `complex` 型別？因為複數在科學計算、工程、物理學、訊號處理等領域十分常見。
4. Python 為什麼沒有內建的有理數型別？因為大部分程式設計工作使用int/float就已經足夠。有理數的需求相對較少，如果每個程式都內建有理數支援，會增加語言與實作的複雜度。
5. 下列三個物件的型別有何不同？

```python
1 --> int (integer)
1.0 --> float (real number)
1+0j --> complex (complex number)
```

## 第三部分：CPython 如何實作這些型別？

除了知道 Python 使用哪些型別表示數學物件之外，

我們也希望了解：

> Python 物件在 CPython 中是如何實作的？

請查閱資料，完成下表。

| 數學概念 | Python 型別 | CPython 實作 | 主要特點 |
|----------|------------|-------------|----------|
| 自然數 | `int` | PyLongObject | 任意精度整數，沒有固定大小限制 |
| 整數 | `int` | PyLongObject | 可表示正數、負數與 0，精度不限 |
| 實數 | `float` | PyFloatObject | 採用 IEEE 754 雙精度浮點數，屬於近似值表示 |
| 複數 | `complex` | PyComplexObject | 同時儲存實部與虛部，可直接進行複數運算 |
| 集合 | `set` | PySetObject | 元素不重複、無順序，支援交集、聯集等集合運算 |
| 命題真假 | `bool` | PyBoolObject | 僅有 True、False 兩種值，且是 int 的子型別 |
| 串列 | `list` | PyListObject | 可變動、有順序，可儲存不同型別的物件 |

## 第六部分：AI 模型中的數值表示

人工智慧模型需要儲存大量參數（Parameters）與權重（Weights）。

請查閱資料並回答：

1. AI 模型中的權重通常使用什麼型別儲存？AI 權重通常使用浮點數，如 FP32、FP16、BF16。
2. 什麼是 FP64？FP64 是 64 位元雙精度浮點數。
3. 什麼是 FP32？FP32 是 32 位元單精度浮點數。
4. 什麼是 FP16？FP16 是 16 位元半精度浮點數。
5. 什麼是 BF16？BF16 是 16 位元 Brain Floating Point 格式，適合 AI 訓練。
6. 為什麼大型 AI 模型不直接使用 Python 的 `int` 作為權重？因為 AI 權重通常包含小數，需要浮點數表示，而 int 只能表示整數，且記憶體效率較差。

## 第八部分：綜合思考

### 思考題一

Python 官方文件指出：

```python
int
```

具有 Unlimited Precision（無限制精度）。

請回答：

1. Unlimited Precision 是什麼意思？Unlimited Precision 指整數沒有固定的位元數限制，可依需求動態增加儲存空間。
2. 它是否代表可以儲存無限大的整數？不是真的無限大，仍受記憶體與系統資源限制。
3. CPython 如何實現這項特性？CPython 使用 PyLongObject 實作 int，並以多個 digit 區塊儲存整數，需要時自動配置更多空間。

---

### 思考題二

比較：

```python
int
```

與

```python
float
```

回答：

1. 為什麼 `int` 幾乎沒有固定上限？int 幾乎沒有固定上限，因為使用任意精度（Arbitrary Precision）設計，會自動擴充儲存空間。
2. 為什麼 `float` 卻存在範圍限制？float 使用固定大小的 IEEE 754 雙精度浮點數，因此有範圍限制。
3. 為什麼 `float` 可能產生誤差？float 可能產生誤差，因為許多十進位小數無法用二進位精確表示，只能儲存近似值。

### 思考題三

比較下列兩種設計理念：

#### Python 的整數

```text
Arbitrary Precision
正確性優先
```

#### AI 模型常用的 FP16、BF16

```text
Limited Precision
效率優先
```

請說明：

1. 兩種設計理念的差異。
Python 整數採用 Arbitrary Precision，重視正確性；FP16、BF16 採用 Limited Precision，重視效率。
2. 各自的優點與缺點。
Arbitrary Precision 的優點是精確且不易溢位，缺點是較慢且耗記憶體；FP16、BF16 的優點是速度快、記憶體需求低，缺點是精度較低。
3. 為什麼大型語言模型通常選擇較低精度的表示方式？
大型語言模型使用 FP16、BF16，因為能降低記憶體需求、提升訓練速度並降低成本。
4. 如果所有 AI 模型都改用 FP64，可能會產生哪些問題？
若全部改用 FP64，記憶體需求會增加、訓練速度變慢、GPU 效率下降，訓練成本也會大幅提高。

### 挑戰題

請用自己的話解釋：

```python
x = 10**1000
```

為什麼 Python 可以正常運作；

而大型 AI 模型卻通常選擇使用：

```text
FP32
FP16
BF16
```

而不是任意精度整數或高精度浮點數？
Python 的 int 採用任意精度（Arbitrary Precision）設計，因此像 10**1000 這樣的超大整數仍然可以正常表示。當數字變大時，Python 會自動配置更多記憶體來儲存資料，因此不容易發生溢位。

大型 AI 模型通常使用 FP32、FP16 或 BF16，而不是任意精度整數或高精度格式。從記憶體使用量來看，FP32、FP16、BF16 都是固定大小，因此能大幅節省記憶體。從計算速度來看，固定精度格式可以由硬體直接加速運算，因此速度較快。從數值精度來看，AI 模型通常不需要完全精確的數值，只需要足夠的精度即可。從 GPU 運算效率來看，現代 GPU 對 FP32、FP16 和 BF16 都有專門的硬體支援，可以提高訓練速度並降低成本。

因此，Python 的 int 重視數值正確性，而 AI 模型更重視記憶體效率、運算速度與 GPU 效能，所以通常選擇使用 FP32、FP16 或 BF16。

- 計算速度
- 數值精度
- GPU 運算效率
