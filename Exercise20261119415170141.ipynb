問題：在 Web Framework（例如 Flask）中，裝飾器Decorator 常被用來將網址路徑（URL Path）與處理函式（Handler Function）建立對應關係。

給定全域字典：
```python
routes = {}
```
請將下面兩個函式
```python
def route(path):
    pass
```
以及：
```python
def dispatch(path):
    pass
```
中的pass代換成新的程式碼，使得下列程式
```python
@route("/")
def home():
    return "Home"

@route("/about")
def about():
    return "About"

@route("/contact")
def contact():
    return "Contact"

while True:

    path = input("URL: ")

    if path == "quit":
        break

    print(dispatch(path))
```
執行結果如下：
```text
URL: /
Home

URL: /about
About

URL: /abc
404 Not Found

URL: quit
```


````python
# 作答區
def route(path):

    def decorator(func):
        # 在這裡可以將路徑與函式綁定起來
        # 例如可以將路徑存入一個全域字典中
        routes[path] = func
        return func
    return decorator
````

````python
# 作答區
def dispatch(path):
    # 在這裡可以根據路徑找到對應的函式並執行
    if path in routes:
        return routes[path]()
    
    return "404 Not Found"
````
