1. 0.1的二進位表示法為何? 使用markdown語法，列出計算過程和理由，並描述看到的現象。
0.1 × 2 = 0.2 → 0
0.2 × 2 = 0.4 → 0
0.4 × 2 = 0.8 → 0
0.8 × 2 = 1.6 → 1
0.6 × 2 = 1.2 → 1
0.2 × 2 = 0.4 → 0
...
 
因此：
0.1 = 0.00011001100110011...
其中 0011 會不斷重複。
 
觀察到的現象是：0.1 在十進位中很簡單，但在二進位中會變成無限循環小數，因此無法被電腦完全精確表示。

2. 如果是無限多位，Python如何解決？提示：IEEE 754、四捨六入五成雙
由於 0.1 的二進位表示法是無限循環小數，因此電腦無法儲存所有位數。Python 使用 IEEE 754 浮點數標準，只保留有限位元數，並利用四捨六入五成雙（Round Half to Even）的方式進行捨入。因此 Python 儲存的是最接近的近似值，而不是完全精確的數值。

3. Python如何知道浮點數的精度？請問執行以下程式碼:
```python
import sys

```
並說明輸出結果的意思。
可以看到 Python 浮點數（float）的相關資訊，例如最大值（max）、最小值（min）、有效位數（dig）與精度（epsilon）等。從輸出結果可以知道，Python 的 float 具有固定的精度限制，並不是無限精度的資料型別，因此無法完全精確表示所有小數。

４. 在數學上0.1+0.2=0.3，可是下面的Python程式的答案好像不是0.3。怎麼解決？
```python
print(0.1+0.2)
```
0.1 與 0.2 無法以二進位精確表示，因此 Python 儲存的是近似值，所以結果可能為 0.30000000000000004。
可使用：
```python
round(0.1 + 0.2, 1)
```
將結果顯示為 0.3。
